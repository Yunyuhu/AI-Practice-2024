## 作業
繪製一個函數圖形

$$(sqrt(cos(x))*cos(200x)+sqrt(abs(x))-0.7)*(4-x*x)^0.01, sqrt(9-x^2), -sqrt(9-x^2) $$

* **np.sqrt()** ：代表平方根，是 "square root" 的縮寫。在數學中，對於一個非負數 $x$，其平方根 $\sqrt{x}$​ 是指一個數字 $y$，使得 $y^2$ = $x$。


* **np.abs()** ：是 Python 中的內置函數，用於計算絕對值。絕對值是指數字的大小，不考慮其正負號。

* **np.cos()** ：是 NumPy 库中的 cos 函数，用于计算给定角度的余弦值。

```python
# 定義函數
def f(x):
    return (np.sqrt(np.cos(x)) * np.cos(200*x) + np.sqrt(np.abs(x)) - 0.7) * (4 - x**2)**0.01
```

- np.linspace(起點，終點，密集度) 是 NumPy 库中的一个函数，用於定義一個等間距的數字。

```python
# 定義 x 範圍
x = np.linspace(-5, 5, 10000)
# 計算函數的值，定義 y 範圍
y = f(x)*1.5
```

- plt.plot() ：是 Matplotlib 庫中的函数，用于繪製二為圖（ex:折線圖、點散圖、柱狀圖等）。

- plt.legend()：是將不同lable的函數標注出它們的對應顏色。

- lw=線條粗細

- lable=標記圖形中每條線所代表的函數或資料的標籤


```python
# 繪製愛心的函數圖形
plt.plot(x, y,lw=5, label='(sqrt(cos(x))*cos(200x)+sqrt(abs(x))-0.7)*(4-x*x)^0.01')
# 繪製由兩個函數組成的圓型圖形
plt.plot(x, np.sqrt(9 - x**2),lw=2, label='sqrt(9-x^2)')
plt.plot(x, -np.sqrt(9 - x**2),lw=2, label='-sqrt(9-x^2)')
# 顯示不同函數所對應的顏色
plt.legend()
```
